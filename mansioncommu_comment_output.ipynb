```
from bs4 import BeautifulSoup
#######
## CopyrightⒸ　2026 SHUICHI HIRAI(平井修一)<@vladesite> All-Rights reserved.  7/10/2026
##①マンションコミュニティの各ページを名前を付けて".html"型式で保存する
##②保存したhtmlファイルをGooglecolabの"📁ファイルストレージ"にアップロードする。./content/
##スクレイピングは出来なくなっているため（Bot防止？CloudflareのWAFを使ってる?)。
##CDN迂回などをすれば対策できるかも。。。仕方ないのでサーバーに保存したhtmlファイルを読み込む
##コメントが多く、複数ページの場合は各ページ毎に保存。　.../res/1-500/ ←※コメントは500単位。以降/501-1000/
########

#ファイル名#
file_path = 'itemprop2.html'

try:
    with open(file_path, 'r', encoding='utf-8') as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, 'html.parser')

    # <p itemprop="commentText"> の内容をすべて抽出
    comments = soup.find_all('p', itemprop='commentText')

    for i, comment in enumerate(comments, 1):
        print(f"Comment {i}: {comment.get_text(strip=True)}")

except FileNotFoundError:
    print(f"Error: {file_path} が見つかりませんでした。")
except Exception as e:
    print(f"An error occurred: {e}")
```
